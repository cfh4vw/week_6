The model used in this decision-making is called a kmeans model. This means that it groups similar players together and helps identify players of interest that could be potentially traded to our team to increase our performance, while not exhausting our budget.

I started by loading in the most recent data from the NBA, including player stats and their salaries. Once I had compiled my data, I noticed that there were duplicate entries for several players, due to being traded mid-season and gaining stats for multiple teams. To deal with these, I decided to only keep the player statistics for the team that they played the most games for. Once I had my data cleaned up, I ran the clustering algorithm to show me 3 different groups of similar players, based on a variety of easy-to-interpret statistics such as games played, field goal percentage, 2- & 3-point percentage, free throw percentage, minutes played, total rebounds, offensive rebound, defensive rebounds, steals, and turnovers. 

From here, I examined a few of these features to see which ones would be best correlated with salary and would best differentiate players based on performance. 






I also examined a few of the variables individually correlated with salary, and decided to use turnovers as one of my indicators since it looked like it had a high variance.



The last visualization I completed was a 3d scatterplot, and is shown in the other notebook. This shows that generally, players who play more, rebound more, and create less turnovers are generally paid more. There were mildly clear clusters shown on this graph, and I identified the one(s) in the middle as our target players.

Once I had all the data visualized, I ran the official KMeans clustering algorithm on our features of interest. I started off using 5 clusters, meaning the algorithm would group players into 1 of 5 groups of similarity based on performance and salary. 

When I ran evaluation metrics on this model, it appeared to account for most of the variance and perform well overall, but other metrics to measure the optimal number of clusters suggest that it could get better. See the elbow plot below, it starts to level out at k=3, meaning that having more than 3 clusters for our data would give us diminishing returns on investment. 



So, using this information, I reran the model with 3 separate clusters. The evaluation metrics told me that these clusters accounted for about 70% of the variance in the data, and that they were overlapping but still reasonable for our purposes. 

Lastly, I created one final visualization that would be a 3D interactive scatter plot, showing all three player statistics, their salary, their assigned cluster, and their name when you hover over the point. I used this visualization to select a few players of interest and demonstrate how to read this model.

- the bottom cluster which represents poor performers that would not add much to our team (represented by a square on the 3d visualization)

- finally, the middle cluster which represents our target trade pool. These players are relatively high performers if you look on the high side, and also likely undervalued on their current team. If we can entice these players with a slightly higher salary and a promise of good playing time, I think we could secure these guys.
